```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# 定义ODE函数
def odefun(t, x):
    return [
        0.8 * x[0] * 0.8 - 0.5 * x[2] - 0.2 * x[0],
        0.5 * x[1] * 0.8 - 0.25 * x[2] - 0.5 * x[1],
        0.5 * x[0] + 0.25 * x[1] - 0.7 * x[4] - 0.05 * x[2],
        0.8 * x[4] * (1 + 0.3 * x[2] / 20),
        0.5 * x[4] + 0.3 * x[4] + 0.5 * x[4] + 0.5 * x[4] - 0.5 * x[4]
    ]

# 初始条件
x0 = [5, 2.5, 0.01, 1, 0.1]
tspan = [0, 24]

# 调用solve_ivp求解
sol = solve_ivp(odefun, tspan, x0, t_eval=np.linspace(tspan[0], tspan[1], 100))

# 绘制结果
plt.plot(sol.t, sol.y[0], 'b-', label='x1')
plt.plot(sol.t, sol.y[1], 'r--', label='x2')
plt.plot(sol.t, sol.y[2], 'g-.', label='x3')
plt.plot(sol.t, sol.y[3], 'k:', label='x4')
plt.plot(sol.t, sol.y[4], 'm-', label='x5')
plt.legend()
plt.xlabel('Time')
plt.ylabel('Solution')
plt.title('Solution of the ODE System')
plt.xlim([0,10])
plt.ylim([-10,20])
plt.show()
```
