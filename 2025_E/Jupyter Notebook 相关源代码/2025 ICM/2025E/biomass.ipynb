```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
from IPython.display import set_matplotlib_formats
set_matplotlib_formats('svg')
# 定义ODE函数
def odefun(t, x):
    return [
        0.1 * x[0] * (1 - x[0] / 300 - 2 * x[2] / 300) - 0.01 * x[0] * x[3] - 0.1 * x[0],
        0.1 * x[1] * (1 - x[1] / 300 - 0.5 * x[0] / 300) - 0.005 * x[0] * x[3] - 0.0001 * x[1],
        0.15 * x[2] * (1 - x[2] / 3) + 0.02 * x[0] * x[2] + 0.02 * x[1] * x[2] - 0.01 * x[3] * x[2],
        0.05 * x[3] * (1 - x[3] / 1) + 0.01 * x[2] * x[3],
        0.05 * x[0] + 0.03 * x[1] + 0.05 * x[2] + 0.05 * x[3] - 0.5 * x[4]
    ]

def odefun2(t, y):
    return [
        0.12 * y[0] * (1 - y[0] / 300 - 1.5 * y[1] / 300) - 0.01 * y[0] * y[2] + 0.001 * y[0],
        0.1 * y[1] * (1 - y[1] / 300 - 2 / 3 * y[0] / 300) - 0.005 * y[0] * y[2] - 0.001 * y[1],
        0.18 * y[2] * (1 - y[2] / 3) + 0.02 * y[0] * y[2] + 0.02 * y[1] * y[2] - 0.01 * y[3] * y[2] - 0.01 * y[4] * y[2] - 0.01,
        0.05 * y[3] * (1 - y[3] / 1 - 0.5 * y[4] / 1) + 0.01 * y[2] * y[3],
        0.05 * y[4] * (1 - y[4] / 1 - 2 * y[3] / 1) + 0.01 * y[2] * y[4]
    ]

# 初始条件
x0 = [80, 40, 15, 5, 3]
y0 = [80, 40, 15, 5, 5]
tspan = [0, 5]

# 调用solve_ivp求解
sol_x = solve_ivp(odefun, tspan, x0, t_eval=np.linspace(tspan[0], tspan[1], 100))
sol_y = solve_ivp(odefun2, tspan, y0, t_eval=np.linspace(tspan[0], tspan[1], 100))

# 绘制结果

plt.figure(figsize=(10,3))
plt.plot(sol_y.t, sol_y.y[2], color='red', label='Pests',linewidth=3,alpha=0.9)
plt.legend()
plt.xlabel('Time (Month)')
plt.ylabel('Biomass')
plt.ylim([0, 50])
plt.grid(True, which='major', color='blue', linestyle='--', linewidth=0.8,alpha=0.4)
plt.grid(True, which='minor', color='lightblue', linestyle=':', linewidth=0.8,alpha=0.4)
plt.minorticks_on()
plt.show()
```

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
from IPython.display import set_matplotlib_formats
set_matplotlib_formats('svg')
# 定义ODE函数
def odefun(t, x):
    return [
        0.1 * x[0] * (1 - x[0] / 300 - 2 * x[1] / 300) - 0.01 * x[0] * x[2] - 0.01,
        0.1 * x[1] * (1 - x[1] / 300 - 0.5 * x[0] / 300) - 0.005 * x[0] * x[2] - 0.0001,
        0.15 * x[2] * (1 - x[2] / 3) + 0.02 * x[0] * x[2] + 0.02 * x[1] * x[2] - 0.01 * x[3] * x[2],
        0.05 * x[3] * (1 - x[3] / 1) + 0.01 * x[2] * x[3],
        0.05 * x[0] + 0.03 * x[1] + 0.05 * x[2] + 0.05 * x[3] - 0.5 * x[4]
    ]

# 初始条件
x0 = [90, 50, 15, 10, 5]
tspan = [0, 10]

# 调用solve_ivp求解
sol = solve_ivp(odefun, tspan, x0, t_eval=np.linspace(tspan[0], tspan[1], 100))
# 设置网格和背景颜色
plt.figure(figsize=(10,3))
# 绘制结果
plt.plot(sol.t, sol.y[0], color='#A8DADC', linestyle='-', label='Crop',linewidth=2)
plt.plot(sol.t, sol.y[1], color='#457B9D', linestyle='--', label='Weed',linewidth=2)
plt.plot(sol.t, sol.y[2], color='#1D3557', linestyle='-.', label='Pest',linewidth=2)
plt.plot(sol.t, sol.y[3], color='#32CD32', linestyle=':', label='Bird',linewidth=2)
plt.plot(sol.t, sol.y[4], color='#E63946', linestyle='-', label='Decomposer',linewidth=2)
plt.legend()
plt.xlabel('Time (Month)')
plt.ylabel('Biomass')
plt.ylim([0, 95])


plt.grid(True, which='major', color='gray', linestyle='--', linewidth=0.5)
plt.grid(True, which='minor', color='lightgray', linestyle=':', linewidth=0.5)
plt.minorticks_on()

plt.show()
```
