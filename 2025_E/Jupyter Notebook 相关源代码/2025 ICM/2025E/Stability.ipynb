```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import set_matplotlib_formats
set_matplotlib_formats('svg')

# 定义物种丰度（百分比）
abundance = np.linspace(0, 100, 500)

# 定义逻辑斯蒂函数
def logistic_function(x, L=1, k=0.1, x0=50):
    return L / (1 + np.exp(-k * (x - x0)))

# 恢复力稳定性（随物种丰度增大而递减）
resilience_stability = logistic_function(abundance, L=1, k=-0.1, x0=50)

# 抵抗力稳定性（随物种丰度增大而递增）
resistance_stability = logistic_function(abundance, L=1, k=0.1, x0=50)

# 创建一个Figure对象和一个子图
fig = plt.figure(figsize=(4,2.5))

# 绘制恢复力稳定性曲线
plt.plot(abundance, resilience_stability, label='Resilience Stability', color='blue',linewidth=3,alpha=0.8)

# 绘制抵抗力稳定性曲线
plt.plot(abundance, resistance_stability, label='Resistance Stability', color='purple',linewidth=3,alpha=0.8)

# 添加图例
plt.legend()
plt.xlabel('Species Biodiversity/Richness (%)')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 定义正态分布的参数
mu1, sigma1 = 0, 0.1  # 第一个正态分布的均值和标准差
mu2, sigma2 = 0, 0.2  # 第二个正态分布的均值和标准差

# 生成数据
x = np.linspace(-1, 1, 1000)
y1 = (1/(sigma1 * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu1) / sigma1)**2)
y2 = (1/(sigma2 * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu2) / sigma2)**2)


# 绘制正态分布
plt.figure(figsize=(10, 3))
plt.plot(x, y1, label='Without Bats',linewidth=4,color='#E8D3A9')
plt.plot(x, y2, label='With Bats',linewidth=4,color='#A5C3CF')

plt.fill_between(x, y1-0.1,y1+0.1,color='#E8D3A9',alpha=0.4)
plt.fill_between(x, y2-0.1,y2+0.1,color='#A5C3CF',alpha=0.4)
plt.legend()
plt.ylabel('Pests\'population(relative ratio)')
plt.xlim=([-1,1])
plt.grid(True)
```
